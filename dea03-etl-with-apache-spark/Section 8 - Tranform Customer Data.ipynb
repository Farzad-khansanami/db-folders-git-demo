```sql
Use catalog gizmobox;
USe schema bronze;
Create OR replace view v_customers as
Select *,
      _metadata.file_path as file_path
From json.`/Volumes/gizmobox/landing/operational_data/customers/`; 
```

```sql
Select distinct *
From v_customers
Where customer_id is not null
order by customer_id
```

```sql
with cte_max as (
Select *,
      row_number() over(partition by customer_id order by created_timestamp desc) as rn
From v_customers
Where customer_id is not null)
Select cast(t.created_timestamp as timestamp) as created_timestamp, 
      cast(t.customer_id as int) as customer_id,
      cast(t.customer_name as string) as customer_name,
      cast(t.date_of_birth as DATE),
      cast(t.email as string) as email,
      cast(t.member_since as date) as member_since,
      CAST(t.telephone as string) as telephone,
      cast(t.file_path as string) as file_path
From cte_max t
Where rn = 1
order by customer_id
```

```sql
CREATE TABLE if not exists gizmobox.silver.customers
AS
with cte_max as (
Select *,
      row_number() over(partition by customer_id order by created_timestamp desc) as rn
From v_customers
Where customer_id is not null)
Select cast(t.created_timestamp as timestamp) as created_timestamp, 
      cast(t.customer_id as int) as customer_id,
      cast(t.customer_name as string) as customer_name,
      cast(t.date_of_birth as DATE),
      cast(t.email as string) as email,
      cast(t.member_since as date) as member_since,
      CAST(t.telephone as string) as telephone,
      cast(t.file_path as string) as file_path
From cte_max t
Where rn = 1
```
