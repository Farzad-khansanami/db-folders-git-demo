```sql
WITH cte AS (
  SELECT *
  FROM read_files(
    '/Volumes/gizmobox/landing/operational_data/addresses/',
    format => 'csv',
    sep => '\t',
    inferSchema => true,
    header => true
  )
)
SELECT *
FROM (
  SELECT customer_id, address_type, address_line_1, city, state, postcode
  FROM cte
) 
PIVOT (
  MAX(address_line_1) as address_line_1,
  MAX(city) as city,
  MAX(state) as state,
  MAX(postcode) as postal_code
  FOR address_type IN ('shipping', 'billing')
);
```
